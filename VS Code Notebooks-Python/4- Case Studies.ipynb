# 🌍 Notebook 4 — Case Studies

---

## What we did

We selected four countries that represent extreme or analytically distinct cases within the dataset and profiled them in depth using only the numbers from our data. No outside knowledge was used — every claim is traceable to a specific variable in the master dataset.

---

## Why these four countries

Each was selected by the data, not chosen in advance:

| Country | Selection criterion |
|---------|-------------------|
| **Mozambique** | Highest measured protection gap in the dataset (100th percentile) |
| **Somalia** | Only country with SP coverage rate of zero |
| **Haiti** | Highest Gini coefficient in the case group (96th percentile globally) |
| **Central African Republic** | Highest CLVI score in the entire dataset (100th percentile) |

---

## Case Study 1 — Mozambique
### *"The highest measured protection gap in the dataset"*

**The numbers:**
- CLVI: **0.695** — 99th percentile globally
- SP coverage: **6.0%** — 4th percentile globally
- Protection gap: **0.635** — 100th percentile (largest in dataset)
- GDP per capita: **$657** — 5th percentile
- Poverty ratio: **81.4%** — 87th percentile
- Health expenditure: **8.5% of GDP** — 74th percentile ← anomaly

**The finding:**
Mozambique spends more on health than most countries at its income level — yet only 6% of its population has any formal social protection. The gap between health spending rank (74th percentile) and SP coverage rank (4th percentile) is not accidental. State spending exists but formal protection delivery infrastructure does not.

> *"A country can spend on health and still fail to build the capability floor. Mozambique's health spending without SP coverage means people can access some services but cannot absorb economic shocks. Capability requires both."*

---

## Case Study 2 — Somalia
### *"Zero measured SP coverage — the floor of the dataset"*

**The numbers:**
- CLVI: **0.596** — 97th percentile globally
- SP coverage: **0.0%** — 1st percentile (lowest in dataset)
- Physical exposure: **0.754** — highest of all five case countries
- Lack of coping capacity: **82.11** — highest in the case group
- GDP per capita: **$630** — 3rd percentile
- Life expectancy: **58.8 years** — 3rd percentile (14.4 years below global mean)
- Poverty ratio: **NO DATA** ← itself a finding

**The finding:**
Somalia's vulnerability is driven almost entirely by physical exposure — not inequality. The crisis is state absence combined with extreme climate exposure. Somalia is legible enough to be counted as absent from the SP system, but not legible enough to have its poverty measured.

> *"The research gap and the protection gap are the same gap. Zero coverage means zero evidence base — there is no variation to analyse, no program to evaluate."*

---

## Case Study 3 — Haiti
### *"The inequality outlier — highest Gini in the case group"*

**The numbers:**
- CLVI: **0.608** — 98th percentile globally
- Gini coefficient: **60.79** — 96th percentile globally
- Top 20% income share: **64.3%**
- Bottom 40% income share: **7.5%** — lowest of all case countries
- Palma ratio: **6.45** — top 10% earn 6.45× the bottom 40%
- SP coverage: **6.6%** — 5th percentile
- Health expenditure: **3.2% of GDP** — 10th percentile (lowest of case group)

**The finding:**
Haiti's vulnerability is driven by socioeconomic inequality more than physical exposure. With a Gini of 60.79, even the 6.6% SP coverage that exists is unlikely to be reaching the bottom deciles. The aggregate coverage figure conceals who is actually being covered.

> *"A Gini of 60.79 at the 96th percentile globally is not a natural outcome. It is the cumulative result of political economy choices. The protection gap in Haiti is inseparable from the inequality that produces it."*

---

## Case Study 4 — Central African Republic
### *"100th percentile CLVI — the apex of the gap"*

**The numbers:**
- CLVI: **0.748** — 100th percentile (highest in entire dataset)
- SP coverage: **NO DATA** ← H5 invisibility case
- GDP per capita: **$516** — 2nd percentile (second lowest in dataset)
- Life expectancy: **57.4 years** — 2nd percentile (15.8 years below global mean)
- Poverty ratio: **71.6%** — 85th percentile
- Gini: **56.24** — 94th percentile
- Health expenditure: **10.7% of GDP** — 90th percentile ← anomaly

**The finding:**
The country with the highest vulnerability score in the entire dataset has no SP coverage data. CAR spends proportionally more on health than most countries — but the high percentage reflects the arithmetic of an extremely low GDP denominator, not generosity. It does not appear in any SP monitoring database.

> *"The country at the 100th percentile for vulnerability is the country with no SP coverage data. The global SP monitoring architecture is most blind precisely where need is highest. The invisibility is not random."*

| Gini | 43.9 | 39.7 | 60.79 | 56.24 |
| Health exp % GDP | 8.5% | 3.7% | 3.2% | 10.7% |
| Primary driver | Physical + poverty | Physical exposure | Inequality | Both equally |

---

## What the case studies show collectively

**No two cases are the same.** Mozambique's problem is delivery infrastructure. Somalia's is state absence. Haiti's is inequality. CAR's is the intersection of everything, compounded by complete data invisibility.

This matters because it means **a single global policy solution will not work.** Effective climate-responsive social protection reform needs to be differentiated — addressing physical exposure in some contexts, inequality in others, and state capacity in others still.

The one thing all four cases share is that they sit at the extreme end of the protection gap, and none of them appear as priorities in the current architecture of international climate finance or social protection reform.

```python
"""CASE STUDIES:

Mozambique — highest protection gap of any country that has SP coverage data in the dataset. 
It has a measured SP coverage rate (so we can calculate the gap precisely) and that gap is the largest in the dataset. 
It is the most unambiguous red zone case where we have actual numbers.

Somalia — SP coverage rate of zero. The only country in the dataset with a recorded zero. That is an extreme data point worth 
examining on its own terms.

Haiti — flagged as a statistical outlier by the z-score analysis in the outlier detection cell. It appeared in the CLVI outliers 
list with z-score above 2, meaning it sits more than two standard deviations above the mean vulnerability score. It is also the
 only Americas country in that outlier list, which makes it regionally distinct from the Africa-dominated extreme end.

Central African Republic — highest CLVI score in the entire dataset (0.748, z-score 3.1). The single most climate-labor vulnerable
 country by our composite index.

Selection logic : highest protection gap with data, lowest SP coverage with data, highest vulnerability outlier in a 
non-African region, highest vulnerability overall. Four different dimensions of extreme, not four arbitrarily chosen countries."""
```

```python
#Cell 1 — Imports & Load

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.dpi'] = 150

master = pd.read_csv('outputs/master_indexed.csv')
```

```python
#Cell 2 — Extract Raw Numbers
case_isos  = ['MOZ', 'SOM', 'HTI', 'CAF']
case_names = {
    'MOZ': 'Mozambique',
    'SOM': 'Somalia',
    'HTI': 'Haiti',
    'CAF': 'Central African Republic'
}

cases = master[master['iso3'].isin(case_isos)].copy()
cases['display_name'] = cases['iso3'].map(case_names)

profile_cols = [
    'display_name','iso3','region_un','income_group',
    'CLVI','physical_exposure','socio_vulnerability',
    'sp_coverage_rate','sp_score','protection_gap',
    'gdp_per_capita','gini','life_expectancy',
    'health_exp_gdp','poverty_ratio',
    'wri_score','exposure','lack_coping','lack_adaptation',
    'gain_score','vulnerability','bottom40_share',
    'top20_share','palma_ratio','unemployment_rate'
]

available = [c for c in profile_cols if c in cases.columns]
cases_display = cases[available].set_index('display_name')

bench_cols = ['CLVI','sp_coverage_rate','gdp_per_capita',
              'life_expectancy','gini','health_exp_gdp',
              'poverty_ratio','wri_score','protection_gap']
global_mean   = master[bench_cols].mean().round(3)
global_median = master[bench_cols].median().round(3)

print("=== CASE STUDY PROFILES (raw numbers) ===")
print(cases_display.T.to_string())

print("\n=== GLOBAL MEANS ===")
print(global_mean.to_string())

print("\n=== GLOBAL MEDIANS ===")
print(global_median.to_string())

print("\n=== GLOBAL PERCENTILE RANKS ===")
for iso, name in case_names.items():
    row = master[master['iso3'] == iso]
    if not row.empty:
        print(f"\n{name}:")
        for col in bench_cols:
            if col in master.columns and pd.notna(row[col].values[0]):
                pct = (master[col] <= row[col].values[0]).mean()
                val = row[col].values[0]
                print(f"  {col:<22}: {val:>8.3f}  ({pct*100:.0f}th percentile)")
            else:
                print(f"  {col:<22}: NO DATA")
```

```python
#Cell 4 — Comparative Visualisation
colors = {
    'Mozambique':                '#d73027',
    'Somalia':                   '#fc8d59',
    'Haiti':                     '#fdae61',
    'Central African Republic':  '#a50026'
}

cases_plot = master[master['iso3'].isin(case_isos)].copy()
cases_plot['display_name'] = cases_plot['iso3'].map(case_names)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle(
    'Red Zone Case Studies — Four Countries the System Has Failed\n'
    'Sen · Duflo · Thunberg · Said · Chomsky',
    fontsize=13, fontweight='bold'
)

# ── 1. Vulnerability vs Protection ───────────────────────────────────────────
metrics = ['CLVI','physical_exposure','socio_vulnerability','sp_score']
labels  = ['CLVI','Physical\nExposure','Socio\nVulnerability','SP Score']
x = np.arange(len(metrics))
width = 0.18

for i, (iso, name) in enumerate(case_names.items()):
    row = cases_plot[cases_plot['iso3'] == iso]
    if not row.empty:
        vals = [row[m].values[0] if pd.notna(row[m].values[0]) else 0
                for m in metrics]
        axes[0,0].bar(x + i*width, vals, width,
                      label=name, color=colors[name], alpha=0.85)

axes[0,0].set_xticks(x + width * 1.5)
axes[0,0].set_xticklabels(labels)
axes[0,0].set_ylabel('Score (0–1)')
axes[0,0].set_ylim(0, 1)
axes[0,0].axhline(master['CLVI'].mean(), color='grey',
                  linestyle=':', alpha=0.6, label='Global CLVI mean')
axes[0,0].set_title('Vulnerability vs Protection\nHigh exposure, near-zero provision',
                     fontweight='bold')
axes[0,0].legend(fontsize=8)

# ── 2. Protection gap in global context ──────────────────────────────────────
axes[0,1].hist(master['protection_gap'].dropna(), bins=30,
               color='#D6E4F0', edgecolor='white', linewidth=0.5,
               label='All countries')
for iso, name in case_names.items():
    row = master[master['iso3'] == iso]
    if not row.empty and pd.notna(row['protection_gap'].values[0]):
        gap = row['protection_gap'].values[0]
        axes[0,1].axvline(gap, color=colors[name], linewidth=2.5,
                          linestyle='--', label=f'{name} ({gap:.2f})')
axes[0,1].set_xlabel('Protection Gap Score')
axes[0,1].set_ylabel('Number of Countries')
axes[0,1].set_title('Where Case Countries Sit\nin the Global Distribution',
                     fontweight='bold')
axes[0,1].legend(fontsize=8)

# ── 3. Socioeconomic context ──────────────────────────────────────────────────
names_bar = list(case_names.values())
gdp_vals  = []
le_vals   = []
for iso in case_names.keys():
    row = master[master['iso3'] == iso]
    gdp_vals.append(row['gdp_per_capita'].values[0]
                    if not row.empty and pd.notna(row['gdp_per_capita'].values[0])
                    else 0)
    le_vals.append(row['life_expectancy'].values[0]
                   if not row.empty and pd.notna(row['life_expectancy'].values[0])
                   else 0)

x2 = np.arange(len(names_bar))
bar_colors = [colors[n] for n in names_bar]
short_names = ['Mozambique','Somalia','Haiti','CAR']

ax2b = axes[1,0].twinx()
axes[1,0].bar(x2 - 0.2, gdp_vals, 0.35,
              color=bar_colors, alpha=0.7, label='GDP per capita (USD)')
ax2b.bar(x2 + 0.2, le_vals, 0.35,
         color=bar_colors, alpha=0.4, label='Life expectancy (years)')
axes[1,0].axhline(master['gdp_per_capita'].mean(), color='navy',
                  linestyle=':', alpha=0.5, label='Global mean GDP')
ax2b.axhline(master['life_expectancy'].mean(), color='green',
             linestyle=':', alpha=0.5, label='Global mean LE')
axes[1,0].set_xticks(x2)
axes[1,0].set_xticklabels(short_names, fontsize=9)
axes[1,0].set_ylabel('GDP per capita (USD)', color='navy')
ax2b.set_ylabel('Life expectancy (years)', color='green')
axes[1,0].set_title('Socioeconomic Context\nGDP and life expectancy vs global mean',
                     fontweight='bold')
axes[1,0].legend(loc='upper left', fontsize=7)
ax2b.legend(loc='upper right', fontsize=7)

# ── 4. Quadrant position ──────────────────────────────────────────────────────
visible = master[master['sp_coverage_rate'].notna()]
axes[1,1].scatter(visible['CLVI'], visible['sp_coverage_rate'],
                  alpha=0.25, color='grey', s=25, label='All countries')

axes[1,1].axhspan(0, 50, xmin=0.5, xmax=1.0,
                  alpha=0.06, color='red')
axes[1,1].text(0.72, 5, 'RED ZONE', color='red',
               fontsize=9, fontweight='bold', alpha=0.5)

for iso, name in case_names.items():
    row = master[master['iso3'] == iso]
    if not row.empty:
        clvi = row['CLVI'].values[0]
        sp   = (row['sp_coverage_rate'].values[0]
                if pd.notna(row['sp_coverage_rate'].values[0]) else 0)
        short = 'CAR' if name == 'Central African Republic' else name
        axes[1,1].scatter(clvi, sp, color=colors[name],
                          s=180, zorder=5,
                          edgecolors='black', linewidth=0.8)
        axes[1,1].annotate(
            short, xy=(clvi, sp),
            xytext=(8, 6), textcoords='offset points',
            fontsize=9, color=colors[name], fontweight='bold'
        )

axes[1,1].axhline(50, color='grey', linestyle='--', alpha=0.3)
axes[1,1].axvline(0.5, color='grey', linestyle='--', alpha=0.3)
axes[1,1].set_xlabel('CLVI Score')
axes[1,1].set_ylabel('SP Coverage Rate (%)')
axes[1,1].set_title('Case Countries on the Quadrant\nAll four in the Red Zone',
                     fontweight='bold')

plt.tight_layout()
plt.savefig('outputs/case_studies.png', dpi=200, bbox_inches='tight')
plt.show()
print("✓ Saved outputs/case_studies.png")
```
